```python
import pandas as pd
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder
```

```python
data=pd.read_csv("car_price_prediction_.csv")
data
```

```python
print(data.columns)
```

```python
le=LabelEncoder()
```

```python

x=data[[ "Year","Mileage", "Engine Size", "Condition"]].copy()
y=data["Price"]
x["Condition"] = le.fit_transform(x["Condition"])
```

```python
model=GaussianNB()
# Convert the continuous 'Price' into discrete bins for classification
y_binned = pd.qcut(y, q=4, labels=False, duplicates='drop')
model.fit(x,y_binned)
```

```python
test_car=pd.DataFrame({"Year":[2018], "Mileage":[223009],"Engine Size":[4.5],"Condition":["Like New"]})
test_car["Condition"] = le.transform(test_car["Condition"])
```

```python
pred=model.predict(test_car)
```

```python
print("predicted price:",pred[0])
```
